```python
# STEP 1: Import libraries
import pandas as pd
import numpy as np

```

```python
# STEP 2: Load dataset
db = pd.read_csv("Salary_Data.csv")
```

```python
# STEP 3: Rename column
db.rename(columns={'YearsExperience': 'Experience'}, inplace=True)
```

```python
# STEP 4: Check missing values
print(db.isnull().sum())
```

```python
# STEP 5: Dataset overview
db.info()
```

```python
# STEP 6: Find rows and columns
print(db.shape)
```

```python
# STEP 7: Statistical overview
print(db.describe())
```

```python
# STEP 8: Visualize using Seaborn
import seaborn as sns
sns.scatterplot(x='Experience', y='Salary', data=db)
```

```python
# STEP 9: Visualize using Matplotlib
import matplotlib.pyplot as plt
plt.scatter(db['Experience'], db['Salary'])
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.show()
```

```python
# STEP 10: Separate input and output
x = db.iloc[:, 0].values.reshape(-1, 1)
y = db.iloc[:, -1].values.reshape(-1, 1)
```

```python
# STEP 11: Split training and testing data
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=1
)
```

```python

# STEP 12: Import Linear Regression
from sklearn.linear_model import LinearRegression
```

```python

# STEP 13: Create model
simple_lr = LinearRegression()
```

```python
# STEP 14: Train model
simple_lr.fit(x_train, y_train)
```

```python
# STEP 15: Display testing data
print("X test values:")
print(x_test)

print("Y test values:")
print(y_test)
```

```python
# STEP 16: Find slope and intercept
print("Slope:", simple_lr.coef_)
print("Intercept:", simple_lr.intercept_)
```

```python
# STEP 17: Predict salary
print("Predicted Salary:")
print(simple_lr.predict([[7.2]]))
```

```python
# STEP 18: Visualize regression line
plt.scatter(x, y)
plt.plot(x, simple_lr.predict(x), color='red')
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Linear Regression')
plt.show()
```

```python
# STEP 19: Evaluate model
from sklearn.metrics import r2_score

r2 = r2_score(y_test, simple_lr.predict(x_test))

print("R2 Score:", r2)
```
